```python
from sklearn.cluster import DBSCAN
from sklearn.datasets import make_moons
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
X,y=make_moons(n_samples=250,noise=0.05) ## create moon-shaped dataset
```

```python
X
```

```python
plt.scatter(X[:,0],X[:,1],c=y)
plt.title("Original make_moons Dataset")
## shows the true labels generated by make_moons
```

```python
##feature scaling(Standard Scaling) 
## Since DBSCAN relies on distance, so scaling is recommended.
from sklearn.preprocessing import StandardScaler
scaler=StandardScaler()
```

```python
X_scaled=scaler.fit_transform(X)
```

```python
X_scaled
```

```python
## DBScan Clustering
dbcan=DBSCAN(eps=0.3) ## Take 0.1, Many points will become noise (-1).
dbcan.fit(X_scaled)
```

```python
labels = dbcan.labels_
labels
```

```python
plt.scatter(X_scaled[:,0],X_scaled[:,1],c=labels)
plt.title("DBSCAN Clustering") ## shows the clusters discovered by the algorithm.

## Conclusion -> DBSCAN successfully identified the natural cluster structure of the dataset.(meaning clusters are non spherical and grouped points correctly based on density)
```
