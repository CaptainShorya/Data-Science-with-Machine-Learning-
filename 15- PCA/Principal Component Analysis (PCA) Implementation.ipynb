```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
import pandas as pd
%matplotlib inline
```

```python
## Load the dataset
from sklearn.datasets import load_breast_cancer
```

```python
cancer_dataset=load_breast_cancer()
```

```python
cancer_dataset.keys()
```

```python
print(cancer_dataset.target_names) ## target are 0 and 1
```

```python
print(cancer_dataset.DESCR)
```

```python
df=pd.DataFrame(cancer_dataset['data'],columns=cancer_dataset['feature_names'])
df.head()
```

```python
## Standardization - Step 1
from sklearn.preprocessing import StandardScaler
scaler=StandardScaler()
```

```python
scaler.fit(df) ## Learn mean and standard devivation and apply z-score over it to make it comparable
```

```python
scaled_data=scaler.transform(df) ## Make Changes thorugh transformation 
```

```python
scaled_data
```

```python
##  Applying PCA Algorithms
## No. PCA is not hit-and-trial at all. It is purely mathematical. 
from sklearn.decomposition import PCA
```

```python
pca=PCA(n_components=2) ## defining our model and reduce our input features to 2 principal components
```

```python
data_pca=pca.fit_transform(scaled_data)
## fit() -> Learning the principal component directions.(Compute covar, eigenvalue and eigenvector and finally top eigenvalues that are required)
## transform() -> projects your data onto those directions. So our dataset becomes: PC1 PC2 instead of original features.(drops perpendiculars from every point onto those axes.Those coordinates becomes your new features)
## Finally, data_pca will have PC1 and PC2.
```

```python
data_pca
```

```python
pca.explained_variance_ ## Provide lambda1 and lambda2 value
```

```python
plt.figure(figsize=(8,6))
plt.scatter(data_pca[:,0],data_pca[:,1],c=cancer_dataset['target'])
plt.xlabel('First principal component')
plt.ylabel('Second Principal Component')
```
