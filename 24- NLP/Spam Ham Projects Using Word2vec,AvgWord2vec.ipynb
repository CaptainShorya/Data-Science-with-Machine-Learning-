```python
import pandas as pd
messages=pd.read_csv('SpamClassifier-master/smsspamcollection/SMSSpamCollection',
                    sep='\t',names=["label","message"])
```

```python
messages
```

```python
from nltk.stem import WordNetLemmatizer
lemmatizer=WordNetLemmatizer()
```

```python
import re
import nltk
nltk.download('stopwords')
```

```python
corpus = []
for i in range(0, len(messages)):
    review = re.sub('[^a-zA-Z]', ' ', messages['message'][i])
    review = review.lower()
    review = review.split() 
    review = [lemmatizer.lemmatize(word) for word in review]
    review = ' '.join(review)
    corpus.append(review)
```

```python
[[i,j,k] for i,j,k in zip(list(map(len,corpus)),corpus, messages['message']) if i<1]
## Since 3 messages became empty after preprocessing, they produced no valid Word2Vec vectors. Instead of returning an empty list (which would have a different shape), we return a zero vector of the same length as all other document vectors using np.zeros(model.vector_size). This ensures every document is represented by a vector of identical dimensions, allowing NumPy and machine learning models to process the data correctly.
```

```python
from nltk import sent_tokenize
from gensim.utils import simple_preprocess
```

```python
words=[]
for sent in corpus:
    sent_token=sent_tokenize(sent)
    print(sent_token)
    for sent in sent_token:
        words.append(simple_preprocess(sent))
```

```python
len(words)
## words has 5569 entries because 3 corpus entries produced no sentence tokens (most likely they became empty strings after preprocessing), so the inner loop never appended anything for them.
## corpus has 5572 entries because every original message was processed.
```

```python
def avg_word2vec(doc):
    # remove out-of-vocabulary words
    #sent = [word for word in doc if word in model.wv.index_to_key]
    #print(sent)
    vectors = [model.wv[word] for word in doc if word in model.wv.index_to_key]

    if len(vectors) == 0:
        return np.zeros(model.vector_size)

    return np.mean(vectors, axis=0)
```
