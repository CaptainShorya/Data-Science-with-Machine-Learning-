```python
corpus = """Hello Welcome, to Krish's Naik NLP tutorials. 
Please do watch the entire course to understand well.
"""
print(corpus)
```

```python
## Tokenisation -> It is the foundational preprocessing step in NLP that breaks raw text into smaller, manageable units called tokens (words, subwords, or characters)
## Paragraph ---> Sentence
from nltk.tokenize import sent_tokenize
documnets = sent_tokenize(corpus)
print(documnets)
type(documnets)

# NLTK uses the Punkt Sentence Tokenizer internally. The tokenizer model is not included with the NLTK package itself, so you must download it separately.
```

```python
for sent in documnets:
    print(sent)
```

```python
## Tokenization
## Paragraph ---> words
## sentence ---> words
from nltk.tokenize import word_tokenize
word_tokenize(corpus)
```

```python
for sent in documnets:
    print(word_tokenize(sent))
```

```python
from nltk.tokenize import wordpunct_tokenize
wordpunct_tokenize(corpus)
```

```python
from nltk.tokenize import TreebankWordTokenizer
## Full Stop(.) will not be treated as a separate word but with respect to last word full stop will be separated.
tokenizer = TreebankWordTokenizer()
tokenizer.tokenize(corpus)
```
