## DBSCAN Clutering

```python
from sklearn.cluster import DBSCAN
import matplotlib.pyplot as plt
%matplotlib inline
from sklearn.datasets import make_circles
```

```python
X, y = make_circles(n_samples=750, factor=0.3, noise=0.1) ## Make a large circle containing a smaller circle in 2d.
X
```

```python
plt.scatter(X[:,0],X[:,1])
```

```python
from sklearn.cluster import DBSCAN
```

```python
dbcan=DBSCAN(eps=0.1)
dbcan.fit(X)
dbcan.labels_ ## Or directly you can use dbcan.fit_predict(X)
```

```python
plt.scatter(X[:,0],X[:,1],c=dbcan.labels_) ## represent cluster labels predicted by DBSCAN
## DBSCAN divided the outer ring into multiple smaller clusters because the chosen eps value was too small to connect the entire ring structure.
## Dark purple points that are scattered here and their are outliers.
```

```python
plt.scatter(X[:,0],X[:,1],c=y) ## represent the actual dataset labels
## The original dataset contains two clusters (an inner circular cluster and an outer ring)
```
