## Isolation Forest Anomaly Detection

```python
import pandas as pd
df=pd.read_csv('healthcare.csv')
df.head()
```

```python
import matplotlib.pyplot as plt
plt.scatter(df.iloc[:,0], df.iloc[:,1])
```

```python
from sklearn.ensemble import IsolationForest
```

```python
clf = IsolationForest(contamination=0.2) ## This tells the model: Approximately 20% of the data points are expected to be anomalies.
clf.fit(df) ## Train the model, learns anomoy structure of the data
predictions = clf.predict(df) ##Predicts whether anomoly or not 
```

```python
predictions ## -1 represent my anomoly points 
```

```python
import numpy as np
index = np.where(predictions < 0)
index
```

```python
x=df.values ## converts the DataFrame into a NumPy array.
x
```

```python
index = np.where(predictions < 0) ## positions (indices) of anomaly points
plt.scatter(df.iloc[:,0], df.iloc[:,1]) ## scatter all the points of the data
plt.scatter(x[index,0], x[index,1], edgecolors="r") ## array[row , column] Index -> row and 0 -> column (Marks the anomoly point)

## Red -> Anomoly Points , Blue -> cluster
```
