## Naive Baye's Machine Learning Algorithms Implementation

```python
from sklearn.datasets import load_iris
```

```python
from sklearn.model_selection import train_test_split
```

```python
X,y=load_iris(return_X_y=True)
```

```python
y
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=0)
```

```python
X_train
```

```python
from sklearn.naive_bayes import GaussianNB
```

```python
gnb=GaussianNB()
```

```python
gnb.fit(X_train,y_train)
```

```python
y_pred=gnb.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score,classification_report,confusion_matrix
```

```python
print(confusion_matrix(y_test,y_pred))
print(accuracy_score(y_test,y_pred))
print(classification_report(y_test,y_pred))
```

```python
import seaborn as sns
```

```python
df = sns.load_dataset('tips')
df.head()
```

```python
df.columns
```

```python
## Make independent features and dependent features
X = df[['total_bill', 'tip', 'sex', 'smoker', 'day', 'size']]
Y = df["time"]
```

```python
Y.value_counts(),X['day'].value_counts(),X['sex'].value_counts(),X['smoker'].value_counts()
```

```python
## Split data into train and test 
from sklearn.model_selection import train_test_split
X_train,X_test,Y_train,Y_test = train_test_split(X,Y,test_size=0.25,random_state=42)
```

```python
## Convert categorical features into numerical features using onehot Encoding
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer


##The ColumnTransformer is a powerful tool in sklearn for applying different preprocessing transformations to specific columns within a dataset. This allows processing based on the nature of each feature.
## for eg. 
"""transformer = ColumnTransformer(transformers=[
    ('t1',SimpleImputer(),['Speed']),
    ('t2',OrdinalEncoder(categories=[['low','high']]),['Average_speed']),
    ('t3',OneHotEncoder(sparse=False,drop='first'),['Gender','City'])
],remainder='passthrough')"""

ct = ColumnTransformer(transformers=[('onehot',OneHotEncoder(drop='first'),['sex', 'smoker', 'day'])]
                       ,remainder='passthrough')
ct
```

```python
X_train_encoded = ct.fit_transform(X_train)
X_train_encoded
```

```python
X_test_encoded = ct.transform(X_test)
X_test_encoded
```

```python
from sklearn.naive_bayes import GaussianNB
gnb = GaussianNB()
```

```python
## scikit-learn automatically handles string targets
gnb.fit(X_train_encoded,Y_train) ##Y_train can be Lunch and Dinner

## Some ML libraries require numeric target: TensorFlow, Pytorch
```

```python
y_pred = gnb.predict(X_test_encoded)
y_pred
```

```python
from sklearn.metrics import accuracy_score,confusion_matrix
print(accuracy_score(Y_test,y_pred))
print(confusion_matrix(Y_test,y_pred))
```
